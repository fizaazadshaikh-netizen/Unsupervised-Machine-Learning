## Task 1- Distance Matrix for Orders

```python
import numpy as np
import pandas as pd

orders_df = pd.DataFrame({
    "ID": ["O1","O2","O3","O4","O5","O6"],
    "time": [25,30,20,35,28,22],
    "dist": [4,6,3,8,5,2]
})

coords = orders_df[["time","dist"]].values
dist_mat = np.zeros((len(coords), len(coords)))

for i in range(len(coords)):
    for j in range(len(coords)):
        dist_mat[i,j] = np.sqrt((coords[i,0]-coords[j,0])**2 + (coords[i,1]-coords[j,1])**2)

dist_df = pd.DataFrame(dist_mat.round(2), index=orders_df["ID"], columns=orders_df["ID"])
print(dist_df)
```

## Task 2- Single Linkage Dendrogram

```python
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import linkage, dendrogram

songs_df = pd.DataFrame({
    "Track": ["S1","S2","S3","S4","S5","S6","S7","S8"],
    "tempo": [90,95,120,125,100,105,140,145],
    "energy": [0.40,0.45,0.70,0.75,0.50,0.55,0.85,0.90]
})

X_vals = songs_df[["tempo","energy"]].values
Z = linkage(X_vals, method="single")

plt.figure(figsize=(10,5))
dendrogram(Z, labels=songs_df["Track"].values)
plt.title("Single Linkage Clustering")
plt.xlabel("Songs")
plt.ylabel("Distance")
plt.show()
```

## Task 3- Comparing Linkage Method

```python
for method in ["single","complete","average"]:
    Z = linkage(X_vals, method=method)
    plt.figure(figsize=(10,5))
    dendrogram(Z, labels=songs_df["Track"].values)
    plt.title(f"{method.capitalize()} Linkage Dendrogram")
    plt.xlabel("Songs")
    plt.ylabel("Distance")
    plt.show()
```
