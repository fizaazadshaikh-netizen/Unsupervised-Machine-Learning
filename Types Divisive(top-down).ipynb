## Task 1- Ward Linkage Dendrogram

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from scipy.cluster.hierarchy import linkage, dendrogram

iris_data = load_iris()
X_vals = iris_data.data

Z_ward = linkage(X_vals, method="ward")

plt.figure(figsize=(12,6))
dendrogram(Z_ward, no_labels=True)
plt.title("Ward Linkage on Iris Dataset")
plt.xlabel("Samples")
plt.ylabel("Distance")
plt.grid(alpha=0.3)
plt.show()
```

## Task 2- Cluster Assignment by Height

```python
from scipy.cluster.hierarchy import fcluster

cut_height = 100
cluster_labels = fcluster(Z_ward, t=cut_height, criterion="distance")

counts = pd.Series(cluster_labels).value_counts().sort_index()

print("Cluster labels:", cluster_labels)
print("\nCluster sizes:")
print(counts)
```
