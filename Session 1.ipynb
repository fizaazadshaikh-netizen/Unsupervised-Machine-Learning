## 1. Random User Dataset

```python
import random

users = []
for user_id in range(1, 11):
    users.append({
        'user_id': user_id,
        'daily_app_opens': random.randint(1, 15),
        'avg_session_duration_min': random.randint(5, 45)
    })

users

#Explanation:  
#This dataset simulates 10 users with two behavioral features:
#Daily App Opens (frequency of usage)
#Average Session Duration (minutes) (engagement level)
```

## 2. Supervised vs Unsupervised Learning

```python
import pandas as pd

comparison = pd.DataFrame({
    'Point': ['Training data', 'Main goal', 'Common methods', 'App example', 'Output'],
    'Supervised Learning': [
        'Uses labeled data',
        'Predict a known target',
        'Classification and regression',
        'Instagram feed ranking based on engagement signals',
        'Predicted class or value'
    ],
    'Unsupervised Learning': [
        'Uses unlabeled data',
        'Discover hidden patterns or groups',
        'Clustering and dimensionality reduction',
        'Spotify grouping listeners or songs by similarity',
        'Clusters, associations, or representations'
    ]
})
comparison

#Explanation:  
#Supervised learning predicts outcomes using labeled data,
#while unsupervised learning discovers hidden structures in unlabeled datasets.
```

## 3. Product Segmentation

```python
products = pd.DataFrame({
    'Product': ['Basic Earphones', 'USB Cable', 'Phone Case', 'Wireless Mouse', 'Keyboard', 'Smartwatch', 'Bluetooth Speaker', 'Gaming Headset'],
    'Price': [799, 499, 699, 1299, 1799, 3499, 2999, 3999],
    'Reviews': [120, 85, 150, 420, 510, 780, 650, 920]
})

products['Cluster'] = ['Cluster 1', 'Cluster 1', 'Cluster 1', 'Cluster 2', 'Cluster 2', 'Cluster 3', 'Cluster 3', 'Cluster 3']
products['Reason'] = [
    'Low price and low popularity', 'Low price and low popularity', 'Low price and low popularity',
    'Medium price and medium popularity', 'Medium price and medium popularity',
    'Higher price and high popularity', 'Higher price and high popularity', 'Higher price and high popularity'
]
products

#Explanation:  
#Products are segmented into three clusters based on price and popularity.
#This helps businesses identify budget, mid‑range, and premium product categories.
```
